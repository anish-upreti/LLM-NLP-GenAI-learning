# Part 1: Understanding Llama 3

### LLaMA 3 (Large Language Model Meta AI 3) is a family of large language models developed by Meta AI (formerly Facebook AI). It’s the successor to LLaMA 2, and as of its release, it's among the top open-weight (but not open-source) models available for research and commercial use.

### 🔍 Key Details about LLaMA 3 (as of 2024):
Release date: April 2024

Developed by: Meta AI

License: Open-weight (free to use with restrictions, not fully open source)

Sizes available:

LLaMA 3–8B

LLaMA 3–70B

### 💡 Notable Features:
Improved performance: LLaMA 3 outperforms GPT-3.5 and is competitive with GPT-4 on many benchmarks.

Multilingual capabilities: Stronger performance across many languages, though still primarily trained in English.

Instruction-tuned variants: Chat-optimized versions (e.g., LLaMA 3–70B-Instruct) are available for applications like chatbots and virtual assistants.

Tokenizer update: Uses a new tokenizer that improves handling of non-English text and code.

## Tuning Llama 3 parameters




### Top-k
Top-k controls how many words the model considers using every time it adds a new word to the response.



### Top-p
Top-p is a similar parameter as top-k, but based on the probability that the model will select certain words instead of the number of words.







## Assigning chat roles
So far, we've explored how to generate responses with Llama and tune their style and length using parameters.

But what if we wanted to refine its tone even more, and maybe even assign it a specific personality? For example, we might be building a customer support chatbot that needs to be used by customers with different expectations.

Some users may need a friendly and conversational assistant, while others expect clear, professional answers.

We can use chat roles to cater to this. With chat roles, we can guide Llama's responses by defining two key roles in our prompts. The system role sets the assistant's personality and style. The user role represents the person asking the question. These roles can be set as part of our Llama call when using create chat completion.
