## Encoding text data
### 🟠 1. One-Hot Encoding

What it is: Represents each word as a binary vector.

How it works:

Each word gets a unique index.

The vector has all 0s except a 1 at the index of the word.

Example (for vocab: ['apple', 'banana', 'orange']):

'apple' → [1, 0, 0]

'banana' → [0, 1, 0]

Pros: Simple and fast.

Cons: Sparse vectors, doesn't capture meaning or similarity between words.



### 🟡 2. Bag of Words (BoW)

What it is: Represents a document by the count of each word in the vocabulary.

How it works:

Builds a vocabulary from all words in the dataset.

Each document becomes a vector of word counts.

Example:

Corpus:

Doc1: "I love machine learning"  
Doc2: "Machine learning is fun"

Vocabulary: [I, love, machine, learning, is, fun]

BoW vectors:

Doc1 → [1, 1, 1, 1, 0, 0]

Doc2 → [0, 0, 1, 1, 1, 1]

Pros: Simple, interpretable.

Cons: Ignores word order and semantics.





### 🟢 3. TF-IDF (Term Frequency–Inverse Document Frequency)

What it is: A weighted version of BoW that downplays common words and highlights rare, important ones.

How it works:

TF = Term Frequency (how often a word appears in a doc).

IDF = Inverse Document Frequency (how rare the word is across all docs).

TF-IDF = TF × IDF

Pros: Highlights informative words, reduces noise from common words.

Cons: Still doesn’t capture word order or deep meaning.





### 🔵 4. Word Embedding

What it is: Dense vector representation of words, capturing semantic meaning.

How it works:

Words are mapped to continuous vectors learned from large corpora (e.g., using Word2Vec, GloVe, FastText).

Similar words have similar vectors.

Example:

'king' - 'man' + 'woman' ≈ 'queen'

Pros: Captures meaning, context, and relationships between words.

Cons: Requires more resources, may need pretraining.
