## Encoding text data
### 🟠 1. One-Hot Encoding

What it is: Represents each word as a binary vector.

How it works:

Each word gets a unique index.

The vector has all 0s except a 1 at the index of the word.

Example (for vocab: ['apple', 'banana', 'orange']):

'apple' → [1, 0, 0]

'banana' → [0, 1, 0]

Pros: Simple and fast.

Cons: Sparse vectors, doesn't capture meaning or similarity between words.



### 🟡 2. Bag of Words (BoW)

What it is: Represents a document by the count of each word in the vocabulary.

How it works:

Builds a vocabulary from all words in the dataset.

Each document becomes a vector of word counts.

Example:

Corpus:

Doc1: "I love machine learning"  
Doc2: "Machine learning is fun"

Vocabulary: [I, love, machine, learning, is, fun]

BoW vectors:

Doc1 → [1, 1, 1, 1, 0, 0]

Doc2 → [0, 0, 1, 1, 1, 1]

Pros: Simple, interpretable.

Cons: Ignores word order and semantics.





### 🟢 3. TF-IDF (Term Frequency–Inverse Document Frequency)

What it is: A weighted version of BoW that downplays common words and highlights rare, important ones.

How it works:

TF = Term Frequency (how often a word appears in a doc).

IDF = Inverse Document Frequency (how rare the word is across all docs).

TF-IDF = TF × IDF

Pros: Highlights informative words, reduces noise from common words.

Cons: Still doesn’t capture word order or deep meaning.





### 🔵 4. Word Embedding

What it is: Dense vector representation of words, capturing semantic meaning.

How it works:

Words are mapped to continuous vectors learned from large corpora (e.g., using Word2Vec, GloVe, FastText).

Similar words have similar vectors.

Example:

'king' - 'man' + 'woman' ≈ 'queen'

Pros: Captures meaning, context, and relationships between words.

Cons: Requires more resources, may need pretraining.

# Part 2: Text classification with PyTorch


### Text classification assigns labels to text, giving meaning to words and sentences. It helps in organizing and giving structure to unstructured data and is crucial in various applications, such as analyzing customer sentiment in reviews, detecting spam emails, or tagging news articles with topics. We'll cover three classification types: binary, multi-class, and multi-label.

###  Binary classification
Binary classification sorts text into two categories, such as spam and not spam, as seen in email spam detection.

### ulti-class classification
Multi-class classification categorizes text into more than two categories. For example, a news article could be classified into one of various categories, like politics, sports, or technology, depending on its content.

### Multi-label classification
In multi-label classification, text can belong to multiple categories simultaneously, unlike multi-class where it belongs to just one category. For example, a book can fit into multiple genres like action, adventure, and fantasy all at the same time.







## CNNs for text classification
### CNN architecture for text
A typical CNN architecture for text classification consists of three layers. The convolutional layer applies filters to the input data to detect patterns. The pooling layer reduces the size of the data while preserving important information. Finally, the fully connected layer uses the previous layer outputs for final predictions.
